```python
nota1 = float(input("Digite a nota 1: "))
nota2 = float(input("Digite a nota 2: "))

media = (nota1 + nota2) / 2

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")

print("Média:", media)
```

```python
numero = int(input("Digite um número: "))

for i in range(1, 11):
    resultado = numero * i
    print(numero, "x", i, "=", resultado)
```

```python
senha = int(input("Digite a senha: "))

while senha != 1234:
    print("Senha incorreta, tente novamente.")
    senha = int(input("Digite a senha: "))

print("Acesso permitido!")
```

```python
secreto = 7
tentativas = 0

numero = int(input("Tente descobrir o número: "))
tentativas = tentativas + 1

while numero != secreto:
    if numero < secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")

    numero = int(input("Tente novamente: "))
    tentativas = tentativas + 1

print("Parabéns! Você acertou.")
print("Tentativas:", tentativas)
```

```python
soma = 0
cont_total = 0
cont_pos = 0
cont_neg = 0

while True:
    numero = int(input("Digite um número: "))

    if numero == 0:
        break

    cont_total = cont_total + 1
    soma = soma + numero

    if numero > 0:
        cont_pos = cont_pos + 1
    else:
        cont_neg = cont_neg + 1

print("Quantidade:", cont_total)
print("Positivos:", cont_pos)
print("Negativos:", cont_neg)
print("Soma:", soma)
```
